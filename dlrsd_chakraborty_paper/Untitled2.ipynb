```python
import os
import shutil
import random
from collections import defaultdict

# Configuration
images_dir = '/home/iiitdmk-param/A.C/Anisha/4th objective/dlrsd/full_images'
masks_dir = '/home/iiitdmk-param/A.C/Anisha/4th objective/dlrsd/full_masks'
output_dir = '/home/iiitdmk-param/A.C/Anisha/4th objective/dlrsd/20%samples'
samples_per_class = 6

# Create output directories
os.makedirs(os.path.join(output_dir, "images"), exist_ok=True)
os.makedirs(os.path.join(output_dir, "masks"), exist_ok=True)

# Group images by class
classes = defaultdict(list)

for img_file in os.listdir(images_dir):
    if img_file.endswith('.png'):
        class_name = img_file.split('_')[0]
        classes[class_name].append(img_file)

# Select and copy files
for class_name, files in classes.items():
    print(f"Selecting {samples_per_class} samples from {class_name}")
    
    if len(files) >= samples_per_class:
        selected = random.sample(files, samples_per_class)
    else:
        selected = files
        print(f"Warning: Only {len(files)} files available for {class_name}")
    
    for file in selected:
        # Copy image
        shutil.copy2(
            os.path.join(images_dir, file),
            os.path.join(output_dir, "images", file)
        )
        # Copy mask (assuming same filename)
        shutil.copy2(
            os.path.join(masks_dir, file), 
            os.path.join(output_dir, "masks", file)
        )

print("Done! Files organized in:", output_dir)
```
