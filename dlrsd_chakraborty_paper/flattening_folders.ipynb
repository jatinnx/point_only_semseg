```python
import os
import shutil
from pathlib import Path

def copy_and_rename_images(source_folder, destination_folder):
    """
    Copy all images from subfolders to a single destination folder
    with renamed files in sequence: subfolder_1, subfolder_2, etc.
    """
    # Create destination folder if it doesn't exist
    Path(destination_folder).mkdir(parents=True, exist_ok=True)
    
    # Supported image extensions
    image_extensions = {'.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff', '.webp'}
    
    # Counter for sequential numbering
    global_counter = 1
    file_mapping = []
    
    # Iterate through all subfolders
    for subfolder in os.listdir(source_folder):
        subfolder_path = os.path.join(source_folder, subfolder)
        
        if os.path.isdir(subfolder_path):
            print(f"Processing folder: {subfolder}")
            
            # Get all image files in the subfolder
            image_files = []
            for file in os.listdir(subfolder_path):
                file_path = os.path.join(subfolder_path, file)
                if os.path.isfile(file_path) and Path(file).suffix.lower() in image_extensions:
                    image_files.append(file)
            
            # Sort files for consistent ordering
            image_files.sort()
            
            # Process each image file
            for i, filename in enumerate(image_files, 1):
                source_path = os.path.join(subfolder_path, filename)
                file_extension = Path(filename).suffix
                
                # Create abbreviated folder name
                if len(subfolder) > 5:
                    folder_abbr = subfolder[:5].lower()
                else:
                    folder_abbr = subfolder.lower()
                
                # Create new filename
                new_filename = f"{folder_abbr}_{global_counter}{file_extension}"
                destination_path = os.path.join(destination_folder, new_filename)
                
                # Copy file
                shutil.copy2(source_path, destination_path)
                file_mapping.append((source_path, destination_path))
                
                print(f"  Copied: {filename} -> {new_filename}")
                global_counter += 1
    
    print(f"\nTotal files copied: {global_counter - 1}")
    return file_mapping

# Usage
if __name__ == "__main__":
    source_folder = "/home/iiitdmk-param/Desktop/Anisha/Images_png"  # Change this
    destination_folder = '/home/iiitdmk-param/A.C/Anisha/4th objective/dlrsd/fu_images'  # Change this
    
    copy_and_rename_images(source_folder, destination_folder)
```

```python
import os
import numpy as np
import cv2
from scipy.ndimage import label
from tqdm import tqdm

# ✅ DLRSD class color mapping
CLASS_COLORS = {
    1: (166, 202, 240),   # airplane
    2: (128, 128, 0),     # bare soil
    3: (0, 0, 128),       # buildings
    4: (255, 0, 0),       # cars
    5: (0, 128, 0),       # chaparral
    6: (128, 0, 0),       # court
    7: (255, 233, 233),   # dock
    8: (160, 160, 164),   # field
    9: (0, 128, 128),     # grass
    10: (90, 87, 255),    # mobile home
    11: (255, 255, 0),    # pavement
    12: (255, 192, 0),    # sand
    13: (0, 0, 255),      # sea
    14: (255, 0, 92),     # ship
    15: (128, 0, 128),    # tanks
    16: (0, 255, 0),      # trees
    17: (0, 255, 255)     # water
}

BACKGROUND_INDEX = 255  # internal label for "ignore"
BACKGROUND_COLOR = (100, 100, 100)  # gray for visualization

def rgb_to_index(mask_rgb):
    index_mask = np.full(mask_rgb.shape[:2], BACKGROUND_INDEX, dtype=np.uint8)
    for class_idx, color in CLASS_COLORS.items():
        match = np.all(mask_rgb == np.array(color), axis=-1)
        index_mask[match] = class_idx
    return index_mask

def index_to_rgb(index_mask):
    rgb_mask = np.zeros((*index_mask.shape, 3), dtype=np.uint8)
    for class_idx, color in CLASS_COLORS.items():
        rgb_mask[index_mask == class_idx] = color
    rgb_mask[index_mask == BACKGROUND_INDEX] = BACKGROUND_COLOR
    return rgb_mask

def generate_point_annotation(mask_rgb, max_points_per_region=5):
    """
    Place multiple points in the eroded interior (not boundary) of each connected region.
    """
    index_mask = rgb_to_index(mask_rgb)
    h, w = index_mask.shape
    point_mask = np.full((h, w), BACKGROUND_INDEX, dtype=np.uint8)

    for cls in CLASS_COLORS.keys():
        binary = (index_mask == cls).astype(np.uint8)
        if np.sum(binary) == 0:
            continue

        labeled, num = label(binary)
        for i in range(1, num + 1):
            region = (labeled == i).astype(np.uint8)

            # Erode region to avoid boundaries
            eroded = cv2.erode(region, np.ones((3, 3), np.uint8), iterations=1)
            eroded_coords = np.column_stack(np.where(eroded))
            if eroded_coords.shape[0] == 0:
                continue  # skip very small regions

            # Centroid
            coords = np.column_stack(np.where(region))
            centroid = coords.mean(axis=0)
            cy, cx = int(round(centroid[0])), int(round(centroid[1]))

            if not eroded[cy, cx]:
                dists = np.sum((eroded_coords - [cy, cx])**2, axis=1)
                cy, cx = eroded_coords[np.argmin(dists)]

            point_mask[cy, cx] = cls

            # Extra random points
            np.random.shuffle(eroded_coords)
            count = 0
            for y, x in eroded_coords:
                if point_mask[y, x] == BACKGROUND_INDEX:
                    point_mask[y, x] = cls
                    count += 1
                if count >= (max_points_per_region - 1):
                    break

    return index_to_rgb(point_mask)

def process_mask_folder_rgb(input_dir, output_dir):
    os.makedirs(output_dir, exist_ok=True)
    mask_files = [f for f in os.listdir(input_dir) if f.lower().endswith(('.png', '.jpg'))]

    for filename in tqdm(mask_files, desc="Generating DLRSD multi-point annotations"):
        input_path = os.path.join(input_dir, filename)
        output_path = os.path.join(output_dir, filename)

        mask_bgr = cv2.imread(input_path)
        mask_rgb = cv2.cvtColor(mask_bgr, cv2.COLOR_BGR2RGB)

        point_mask_rgb = generate_point_annotation(mask_rgb)

        point_mask_bgr = cv2.cvtColor(point_mask_rgb, cv2.COLOR_RGB2BGR)
        cv2.imwrite(output_path, point_mask_bgr)

# ✅ Example usage
if __name__ == "__main__":
    input_rgb_mask_dir = "/home/iiitdmk-param/A.C/Anisha/4th objective/dlrsd/full_masks"
    output_point_mask_dir = "/home/iiitdmk-param/A.C/Anisha/4th objective/dlrsd/full_point_masks"
    process_mask_folder_rgb(input_rgb_mask_dir, output_point_mask_dir)
```
